```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('marble_factory_sales_5000.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df["Discount_pct"] = df["Discount_pct"].fillna(0)
df.isnull().sum()
```

```python
df['DeliveryDays'] = df['DeliveryDays'].fillna(df['DeliveryDays'].median())
```

```python
plt.boxplot(df['DeliveryDays'])
plt.title('Outliers check')
plt.ylabel('Days')
plt.show()
```

```python
df.duplicated().sum()
```

```python
df = df.drop_duplicates()
df.duplicated().sum()
```

```python
df.columns
df.columns.to_list()
```

```python
df['Category'].value_counts()
```

```python
df['Category'].value_counts().plot(kind='bar')
plt.title('Category')
plt.xlabel('Category')
plt.ylabel('Count')
plt.show
```

```python
df['MarbleType'].value_counts()
```

```python
df['MarbleType'].value_counts().plot(kind='bar')
plt.title('MarbleType')
plt.xlabel('MarbleType')
plt.ylabel('Count')
plt.show
```

```python
df['Finish'].value_counts()
```

```python
df['Finish'].value_counts().plot(kind='bar')
plt.title('Finish')
plt.xlabel('Finish')
plt.ylabel('Count')
plt.show
```

```python
df['Thickness_mm'].value_counts()
```

```python
df['Thickness_mm'].value_counts().plot(kind='bar')
plt.title('Thickness_mm')
plt.xlabel('Thickness_mm')
plt.ylabel('Count')
plt.show
```

```python
df[['Quantity_sqft', 'UnitPrice_PKR', 'Discount_pct', 'TotalAmount_PKR', 'DeliveryDays']].describe()
```

```python
plt.boxplot(df['TotalAmount_PKR'])
plt.title('TotalAmount_PKR Outliers Check')
plt.show()
```

```python
plt.boxplot(df['Quantity_sqft'])
plt.title('Quantity_sqft Outliers Check')
plt.ylabel('Sqft')
plt.show()
```

```python
category_reveune = df.groupby('Category')['TotalAmount_PKR'].sum()
category_reveune.plot(kind='bar')
plt.title('Category Revenue')
plt.xlabel('Category')
plt.ylabel('TotalAmount_PKR')
plt.show()
```

```python
category_count = df.groupby('Category')['TotalAmount_PKR'].count()
category_count.plot(kind='bar')
plt.title('Number of Orders by Category')
plt.xlabel('Category')
plt.ylabel('Order Count')
plt.show()
```

```python
city_revenue = df.groupby('City')['TotalAmount_PKR'].sum().sort_values(ascending=False)
city_revenue.plot(kind='bar')
plt.title('Total Revenue by City')
plt.xlabel('City')
plt.ylabel('Total Revenue (PKR)')
plt.show()
```

```python
city_revenue = df.groupby('City')['TotalAmount_PKR'].count().sort_values(ascending=False)
city_revenue.plot(kind='bar')
plt.title('Total Revenue by City')
plt.xlabel('City')
plt.ylabel('Total Revenue (PKR)')
plt.show()
```

```python
plt.scatter(df['Quantity_sqft'], df['Discount_pct'])
plt.title('Quantity vs Discount')
plt.xlabel('Quantity (sqft)')
plt.ylabel('Discount (%)')
plt.show()
```

```python
numeric_cols = df[['Quantity_sqft', 'UnitPrice_PKR', 'Discount_pct', 'TotalAmount_PKR', 'DeliveryDays', 'Thickness_mm']]
correlation = numeric_cols.corr()

sns.heatmap(correlation, annot=True, cmap='coolwarm')
plt.title('Correlation Heatmap')
plt.show()
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'])
df['OrderDate'].dtype
```

df['Month'] = df['OrderDate'].dt.month

```python
df.groupby('Month')['TotalAmount_PKR'].sum().plot(kind='line', marker='o')
plt.title('Monthly Revenue Trend')
plt.xlabel('Month')
plt.ylabel('Total Revenue (PKR)')
plt.show()
```
